```python
import pandas as pds
import matplotlib.pyplot as plt
file = ('SOCR-HeightWeight.csv')
data = pds.read_csv(file)
q1H = data['Height(Inches)'].quantile(0.25).round(2)
q2H = data['Height(Inches)'].quantile(0.5).round(2)
q3H = data['Height(Inches)'].quantile(0.75).round(2)
iqrH = q3H-q1H
minimumH = (q1H- (1.5*iqrH)).round(2)
maximumH = (q3H+(1.5*iqrH)).round(2)
q1W = data['Weight(Pounds)'].quantile(0.25).round(2)
q2W = data['Weight(Pounds)'].quantile(0.5).round(2)
q3W = data['Weight(Pounds)'].quantile(0.75).round(2)
iqrW = q3W-q1W
minimumW = (q1W- (1.5*iqrW)).round(2)
maximumW = (q3W+(1.5*iqrW)).round(2)
print("Height Dataset:")
fig = plt.figure(figsize =(12, 9))
plt.boxplot(data['Height(Inches)'], orientation = 'horizontal')
plt.show()
print ('The minimum is', minimumH, 'inches')
print ('Quartile 1 is', q1H, 'inches')
print ('Quartile 2 is', q2H, 'inches')
print ('Quartile 3 is', q3H, 'inches')
print ('The maximum is', maximumH, 'inches')
print ('\n\nWeight Dataset:')
fig = plt.figure(figsize =(12, 9))
plt.boxplot(data['Weight(Pounds)'], orientation='horizontal')
plt.show()
print ('The minimum is', minimumW, 'pounds')
print ('Quartile 1 is', q1W, 'pounds')
print ('Quartile 2 is', q2W, 'pounds')
print ('Quartile 3 is', q3W, 'pounds')
print ('The maximum is', maximumW, 'pounds')

```
